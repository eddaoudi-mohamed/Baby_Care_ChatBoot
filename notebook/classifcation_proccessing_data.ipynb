```python
import pandas as pd

df1 = pd.read_csv('../Data/matinf_english (1).csv')
df2 = pd.read_csv('../Data/matinf_english (2).csv')
df3 = pd.read_csv('../Data/matinf_english.csv')

df = pd.concat([df1, df2, df3], ignore_index=True)

```

```python
df['text'] = df['question_en'].fillna('') + ' ' + df['description_en'].fillna('')

# Create 'label' column with value 1
df['label'] = 1

# Keep only the needed columns
df_final = df[['text', 'label']]

df_final = df[['text', 'label']]

```

```python
import pandas as pd

# Load the dataset
dff = pd.read_json('../Data/News_Category_Dataset_v3.json', lines=True)

# Keep only the desired columns
dff = dff[['headline', 'category', 'short_description']]

# Group by 'category' and sample 135 rows per category
sampled_df = dff.groupby('category', group_keys=False).apply(lambda x: x.sample(n=135, random_state=42))

# Optional: Reset index
sampled_df = sampled_df.reset_index(drop=True)
sampled_df['text'] = sampled_df['headline'] + " " + sampled_df['short_description']
sampled_df['label'] = 0

sampled_df.to_csv('news_sample_135_per_category.csv', index=False)

final_df = sampled_df[['text', 'label']]


```

```python
import pandas as pd

# Concatenate the two dataframes
combined_df = pd.concat([df_final, final_df], ignore_index=True)

# Shuffle the combined dataframe
shuffled_df = combined_df.sample(frac=1, random_state=42).reset_index(drop=True)

# Show the result
print(shuffled_df.head())
```

```python
# Save the shuffled dataframe to a CSV file
shuffled_df.to_csv('../Data/combined_dataset.csv', index=False)
```
